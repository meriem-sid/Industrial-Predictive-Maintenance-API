```python
from sklearn.model_selection import GroupKFold, KFold, cross_validate
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from pdm.data import load_cmapss, load_rul
```

```python
def evaluate(X, y, groups, cv):
    pipe = make_pipeline(StandardScaler(), Ridge(alpha=1.0))   # scaler refit in every fold
    res = cross_validate(pipe, X, y, groups=groups, cv=cv,
                         scoring=["neg_mean_absolute_error", "neg_root_mean_squared_error"])
    return {"MAE": -res["test_neg_mean_absolute_error"].mean(),
            "RMSE": -res["test_neg_root_mean_squared_error"].mean(),
            "RMSE_std": res["test_neg_root_mean_squared_error"].std()}
```
